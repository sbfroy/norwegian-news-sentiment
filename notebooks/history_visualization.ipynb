```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

base_dir = Path.cwd().parent

# Training history of the fine-tuned NB-BERT model
df = pd.read_csv(base_dir / 'logs' / 'run_20250123_162502' / 'history.csv')
epochs = df.index + 1
```

```python
plt.figure(figsize=(10, 6))

# RMSE
plt.subplot(2, 1, 1)
plt.plot(epochs, df['train_rmse'], label='Train RMSE')
plt.plot(epochs, df['val_rmse'], label='Validation RMSE')
plt.ylabel('RMSE')
plt.legend()

# R^2
plt.subplot(2, 1, 2)
plt.plot(epochs, df['train_r2'], label='Train R^2')
plt.plot(epochs, df['val_r2'], label='Validation R^2')
plt.xlabel('Epoch')
plt.ylabel('R^2')
plt.legend()

plt.tight_layout()
plt.show()
```
